```python
# Generate data

import csv
import random

# Số lượng bản ghi muốn tạo
NUM_DRIVERS = 10
NUM_CUSTOMERS = 8
NUM_RIDES = 12
```

```python
# Một số tên mẫu
driver_names = ["Cường", "Dũng", "Hoàng", "Minh", "Nam", "Phúc", "Tùng"]
customer_names = ["Hoa", "Lan", "Mai", "Đào", "Vy", "Ngọc", "Huyền", "Trang"]
locations = ["Q1", "Q2", "Q3", "Q4", "Q5", "Q6", "Q7", "Q8", "Q9", "Q10", "Q11", "Q12"]
```

```python
# --- Tạo file drivers.csv ---

drivers = [["ID", "Name", "Rating", "x", "y"]]
for i in range(1, NUM_DRIVERS + 1):
    name = random.choice(driver_names)
    rating = round(random.uniform(4.0, 5.0), 1)
    x, y = random.randint(1, 10), random.randint(1, 10)
    drivers.append([i, name, rating, x, y])

with open("drivers.csv", "w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows(drivers)
```

```python
# --- Tạo file customers.csv ---

customers = [["ID", "Name", "Location", "x", "y"]]
for i in range(1, NUM_CUSTOMERS + 1):
    name = random.choice(customer_names)
    location = random.choice(locations)
    x, y = random.randint(1, 10), random.randint(1, 100)
    customers.append([i, name, location, x, y])

with open("customers.csv", "w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows(customers)
```

```python
# --- Tạo file rides.csv ---

rides = [["RideID", "CustomerID", "DriverID", "Distance(km)", "Fare(VND)"]]
for i in range(1, NUM_RIDES + 1):
    customer_id = f"C{random.randint(1, NUM_CUSTOMERS)}"
    driver_id = random.randint(1, NUM_DRIVERS)
    distance = round(random.uniform(1.0, 10.0), 1)
    fare = int(distance * random.randint(6000, 9000))  # tính giá ngẫu nhiên
    rides.append([i, customer_id, driver_id, distance, fare])

with open("rides.csv", "w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows(rides)
```
